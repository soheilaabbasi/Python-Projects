```python
# Cashing as Testing(practicing)

from functools import cache
```

```python
@cache
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
from cachetools import cached, TTLCache
```

```python
ttl_cache = TTLCache(maxsize=100, ttl=10)   

# ttl is based on 'seconds' and after 300s it is invalidating a value and calculates it again.
```

```python
from functools import cache, lru_cache
import time

# @lru_cache(maxsize=10000)
@cached(ttl_cache)     # * we give ttl_cache object, that is valid only for 10s.
def get_data(key):
    time.sleep(2)
    return f'Data for {key}'
```
