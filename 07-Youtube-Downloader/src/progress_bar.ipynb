```python
names = [
    'Mina',
    'Nika',
    'Hamidreza',
    'Niloufar',
    'Elaheh',
]
```

```python
import time
from tqdm import tqdm

# for name in tqdm(names, desc='Processing names...'):
#     time.sleep(1)

# OR

pbar = tqdm(names)
for name in pbar:
    pbar.set_description(f'processing {name}')
    print(pbar.n)
    pbar.update(1)
    time.sleep(1)
```

```python
for i in tqdm(range(100)):
    time.sleep(0.1)
```

```python
from tqdm import trange

for i in trange(100):
    time.sleep(0.1)
```
