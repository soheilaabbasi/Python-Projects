- ContactBook
- Methods:
    - add_contact (create)
    - view_contacts (read)
    - update_contact (update)
    - delete_contact (delete)

```python
{
    'Ali':{
        'phone': '1234567890',
        'email': 'hejazizo@gmail.com'
    },
    'Reza':{
        'phone': '0987654321',
        'email': 'reza@gmail.com'
    }
}
```

```python
from collections import defaultdict

class ContactBook():
    def __init__(self):
        self.contacts = defaultdict(dict)

    def add_contact(self, name, phone, email = None):
        if name in self.contacts:
            print("Contact already exists!")
            return
        else:
            self.contacts[name]['phone'] = phone
            self.contacts[name]['email'] = email
```

```python
book = ContactBook()
```

```python
book.add_contact(
    name = 'Ali',
    phone = '123-456-78',
    email = 'hejazizo@ualberta.ca'
)
```

```python
book.view_contact()
```

```python
book.update_contact(
    name = 'Ali',
    # email = 'newemail@gmail.com',
    phone = '1234345254'

)
```
