- __Status codes__:
- 200 → (OK) request is successful
- 403 → Forbidden
- 404 → not found
- 500 → Internal Server Error

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'
}


requests.get('https://www.google.com', headers=headers)
```

```python
# with this Function we can check if a Website is available or not.

def check_site_availability(url):
    try:
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'
        }
        response = requests.get(url, headers=headers, timeout=5)
        if response.status_code == 200:
            print(url, "→", "True")
            return True
        else:
            print(url, "→", "False")
            return False
    except Exception as e:
        print(url, "→", "e")
        return False
```

```python
check_site_availability('https://www.hdajkfsoiiof.com')
```

```python
websites_to_check = [
    "https://www.google.com",
    "https://www.github.com",
    "https://stackoverflow.com",
    "https://www.python.org",
    "https://www.nonexistentwebsites.example"
]
```

```python
for url in websites_to_check:
    print(check_site_availability(url))
```

```python
import threading
```

```python
check_site_availability('https://www.google.com')
```

```python
websites_to_check = [
    "https://www.google.com",
    "https://www.github.com",
    "https://stackoverflow.com",
    "https://www.python.org",
    "https://www.nonexistentwebsites.example"
]
```

- Thread
   │
   ├── target → کدام تابع را اجرا کنم؟
   │
   └── args   → چه ورودی‌هایی به آن تابع بدهم؟

```python
threads = []
for url in websites_to_check:
    thread = threading.Thread(            
        target = check_site_availability,  # target → کدام تابع را اجرا کنم؟
        args = (url,)                      # args → چه ورودی‌هایی به آن تابع بدهم؟
    )                                      #args parameter should be a tuple.
                                           #so, we should put a ',' in front of 'url' to be a touple
    threads.append(thread)
    thread.start()
    
```

```python
for thread in threads:
    thread.join()
```

                      ↓

                      ↓
